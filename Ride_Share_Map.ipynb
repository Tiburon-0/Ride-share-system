```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np

# Create a map using the Barabasi network model
denver_map = nx.barabasi_albert_graph(n=100, m=2, seed=42)

if nx.is_connected(denver_map) == True:
    connected = nx.is_connected(denver_map)
else:
    disconnected = nx.is_connected(denver_map)

# Check connectivity
if connected:
    print(f"This map is connected.\nNodes: {denver_map.nodes()}\nNumber of nodes: {denver_map.number_of_nodes()}\nEdges: {denver_map.edges()}\nNumber of Edges: {denver_map.number_of_edges()}")
elif disconnected:
    print(f"This map is disconnected. Generate map again.")
```
